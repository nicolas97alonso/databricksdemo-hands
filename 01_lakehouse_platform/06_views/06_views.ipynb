# 06 - Views

This notebook demonstrates standard views, session-scoped temporary views, and cluster-scoped global temporary views.

### Key Concepts Covered:
* Standard Views (`CREATE VIEW`)
* Session Temporary Views (`CREATE TEMP VIEW`)
* Global Temporary Views (`CREATE GLOBAL TEMP VIEW`)
* Namespace prefixes for global temp views (`global_temp.`)


## Step 1: Switch Context to aura_raw Schema


```python
%sql
USE CATALOG demoworkspace2;
USE SCHEMA aura_raw;
```

## Step 2: Create a Standard View (Stored in Schema Metadata)


```python
%sql
CREATE OR REPLACE VIEW view_electronics_sales AS
SELECT * FROM store_sales
WHERE product_category = 'Electronics';
```

## Step 3: Query Standard View


```python
%sql
SELECT * FROM view_electronics_sales;
```

## Step 4: Create a Temporary View (Session Scoped)


```python
%sql
CREATE OR REPLACE TEMP VIEW temp_view_groceries_sales AS
SELECT * FROM store_sales
WHERE product_category = 'Groceries';
```

## Step 5: Query Session Temporary View


```python
%sql
SELECT * FROM temp_view_groceries_sales;
```

## Step 6: Create a Global Temporary View (Shared Across Notebooks/Sessions on Same Cluster)


```python
%sql
CREATE OR REPLACE GLOBAL TEMP VIEW global_temp_sports_sales AS
SELECT * FROM store_sales
WHERE product_category = 'Sports';
```

## Step 7: Query Global Temporary View (Must Use global_temp Prefix)


```python
%sql
SELECT * FROM global_temp.global_temp_sports_sales;
```

## Step 8: View Namespace Verification


```python
%sql
SHOW TABLES IN global_temp;
```

## Step 9: Cleanup View Objects


```python
%sql
DROP VIEW IF EXISTS view_electronics_sales;
DROP VIEW IF EXISTS temp_view_groceries_sales;
DROP VIEW IF EXISTS global_temp.global_temp_sports_sales;
```
