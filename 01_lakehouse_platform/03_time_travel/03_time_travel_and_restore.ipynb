# 03 - Time Travel and Restore

This notebook explores Delta Lake's transaction log to query previous table states (time travel) and restore deleted data.

### Key Concepts Covered:
* Querying older versions (`VERSION AS OF`)
* Dropping all records from a table
* Restoring a table to a specific version (`RESTORE TABLE`)


## Step 1: Query Current State


```python
%sql
SELECT * FROM demoworkspace2.aura_raw.store_sales;
```

## Step 2: Delete All Table Records


```python
%sql
DELETE FROM demoworkspace2.aura_raw.store_sales;
```

## Step 3: Verify Table is Empty


```python
%sql
SELECT COUNT(*) FROM demoworkspace2.aura_raw.store_sales;
```

## Step 4: Inspect History to Find Version to Restore


```python
%sql
DESCRIBE HISTORY demoworkspace2.aura_raw.store_sales;
```

## Step 5: Restore Table to Pre-delete State (Version 3)


```python
%sql
RESTORE TABLE demoworkspace2.aura_raw.store_sales VERSION AS OF 3;
```

## Step 6: Verify Data Restoration


```python
%sql
SELECT COUNT(*) FROM demoworkspace2.aura_raw.store_sales;
```
