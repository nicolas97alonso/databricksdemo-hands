# 02 - Delta Table Basics

This notebook creates a managed Delta table, inserts mock data, and performs updates/deletes while tracking table history.

### Key Concepts Covered:
* Managed Delta Table Creation
* Inserting data
* Checking table metadata and history (`DESCRIBE EXTENDED`, `DESCRIBE HISTORY`)
* Table updates and deletions


## Step 1: Set Namespace Context


```python
%sql
USE CATALOG demoworkspace2;
USE SCHEMA aura_raw;
```

## Step 2: Create Managed Delta Table


```python
%sql
DROP TABLE IF EXISTS store_sales;

CREATE TABLE store_sales (
    sale_id INTEGER,
    store_id STRING,
    product_category STRING,
    quantity INTEGER,
    amount DOUBLE,
    sale_timestamp TIMESTAMP
);
```

## Step 4: Describe Table Details


```python
%sql
DESCRIBE TABLE EXTENDED store_sales;
```

## Step 5: Check Table History


```python
%sql
DESCRIBE HISTORY store_sales;
```

## Step 6: Update Table Row


```python
%sql
UPDATE store_sales
SET quantity = 5
WHERE sale_id = 1;
```

## Step 7: Delete Table Row


```python
%sql
DELETE FROM store_sales
WHERE sale_id = 10;
```

## Step 8: Observe Resulting History


```python
%sql
DESCRIBE HISTORY store_sales;
```
