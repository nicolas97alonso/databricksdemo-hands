# 01 - Setup and Schemas

This notebook initializes the catalog and schemas needed for the training.

### Key Concepts Covered:
* Catalog selection (`USE CATALOG`)
* Schema creation (`CREATE SCHEMA`)
* Switching working schemas (`USE SCHEMA`)


## Step 1: Switch to Catalog


```python
%sql
USE CATALOG demoworkspace2;
```

## Step 2: Create raw and gold schemas


```python
%sql
CREATE SCHEMA IF NOT EXISTS aura_raw;
CREATE SCHEMA IF NOT EXISTS aura_gold;
```

## Step 3: Verify Schemas


```python
%sql
SHOW SCHEMAS;
```

## Step 4: Set Active Working Schema


```python
%sql
USE SCHEMA aura_raw;
```

## Step 5: Describe Active Schema Metadata


```python
%sql
DESCRIBE SCHEMA EXTENDED aura_raw;
```
